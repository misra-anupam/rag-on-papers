# Structure aware

The document sections are first isolated, and if their token count exceed a threshold(e.g. 512), it is resursively split using the recursive text spliter.

## Recursive text splitter

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
chunker = RecursiveCharacterTextSplitter(separators="\n\n")

chunks = chunker.split_text(parsed_paper)
```
